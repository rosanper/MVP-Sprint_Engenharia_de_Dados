# Criação das tabelas Bronze

Nesse notebook será consumido os arquivos brutos salvos no volume da camada raw e transformados em tabelas, criando assim a camada bronze

## Deletar as tabelas caso existam

```python
%sql
DROP TABLE IF EXISTS mvp_engenharia_dados.bronze.cid
```

```python
%sql
DROP TABLE IF EXISTS mvp_engenharia_dados.bronze.ibge
```

```python
%sql
DROP TABLE IF EXISTS mvp_engenharia_dados.bronze.sim
```

## Leitura dos arquivos

```python
# Leitura do arquivo do ibge
cid = spark.read.format("csv").option("header", "true").load("/Volumes/mvp_engenharia_dados/raw/mvp_raw/cid_raw.csv")
display(cid)
```

```python
# Leitura do arquivo do ibge
ibge = spark.read.format("csv").option("header", "true").load("/Volumes/mvp_engenharia_dados/raw/mvp_raw/ibge_raw.csv")
display(ibge)
```

```python
# Leitura do arquivo do ibge
sim = spark.read.format("csv").option("header", "true").load("/Volumes/mvp_engenharia_dados/raw/mvp_raw/sim_raw.csv")
display(sim)
```

## Salvando os datframes como tabelas da camada bronze

```python
# # Tabela cid
cid.write.mode("overwrite").saveAsTable("mvp_engenharia_dados.bronze.cid")

# Tabela ibge
ibge.write.mode("overwrite").saveAsTable("mvp_engenharia_dados.bronze.ibge")

# Tabela sim
sim.write.mode("overwrite").saveAsTable("mvp_engenharia_dados.bronze.sim")
```
